```python
print(df.columns.duplicated())
```

```python
categorical_columns = df.select_dtypes(include="object").columns

for col in categorical_columns:
    print("\n", col)
    print(df[col].unique())
```

```python
for col in categorical_columns:
    df[col] = df[col].astype(str).str.strip()
```

```python
df["Driver Gender"] = df["Driver Gender"].str.strip().str.title()
```

```python
print(df["Driver Gender"].unique())
```

```python
df["Weather Conditions"] = (
    df["Weather Conditions"].str.strip().str.title()
)
```

```python
df["Road Condition"] = (
    df["Road Condition"].str.strip().str.title()
)
```

```python
df["Month"] = df["Month"].str.strip().str.title()
```

```python
df["Day of Week"] = df["Day of Week"].str.strip().str.title()
```

```python
df["Time of Day"] = pd.to_datetime(
    df["Time of Day"],
    format="%H:%M",
    errors="coerce"
)
```

```python
print(df["Time of Day"].head())
```

```python
df["Time of Day"] = df["Time of Day"].dt.strftime("%H:%M")
```

```python
numeric_columns = [
    "Year",
    "Number of Vehicles Involved",
    "Number of Casualties",
    "Number of Fatalities",
    "Speed Limit (km/h)",
    "Driver Age"
]

for col in numeric_columns:
    print("\n", col)
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
```
